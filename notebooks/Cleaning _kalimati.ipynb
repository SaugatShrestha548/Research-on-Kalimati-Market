```r
kalimati_csv = read.csv("C:/Users/BROTHER COMPUTER/Desktop/research/Raw_Data/kalimati_combined.csv")
sum(is.na(kalimati_csv))
unique_names <- unique(kalimati_csv$commodity)
commodities_to_search <- c("onion","potato","tomato","cabbage","garlic","cauliflower")
search <- paste(commodities_to_search, collapse = "|")
unique_names[grepl(search, unique_names, ignore.case = TRUE)]
```

```r
#we are focusing in following commodiries:"Onion Dry (Indian)", "Garlic Dry Chinese","Tomato Big(Indian)", "Potato Red", "Cabbage(Local)", "Tomato Big(Nepali)"
choosen_produces <- c("Onion Dry (Indian)", "Garlic Dry Chinese",
                           "Tomato Big(Indian)", "Potato Red",
                           "Cabbage(Local)", "Tomato Big(Nepali)")
filtered_data<- kalimati_csv %>% 
                    filter(commodity %in% choosen_produces)
names(filtered_data)
mutate(Date = ymd(filtered_data$Date))
monthly_average <-filtered_data %>% 
                    mutate(Year = year(Date),
                           Month = month(Date, label = TRUE,abbr = TRUE)) %>%
                    group_by(commodity,Year, Month)%>%
summarize(
    monthly_average = mean(Average, na.rm = TRUE),
    .groups = "drop"
)%>%
    arrange(commodity, Year, Month) 
                    
write.csv(monthly_average, "C:/Users/BROTHER COMPUTER/Desktop/research/Cleaned_Data/kalimati_filtered.csv" )
  
```
