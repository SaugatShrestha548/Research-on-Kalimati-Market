```r
library(dplyr)
library(tidyverse)
library(lubridate)
```

```r
master_csv = read_csv("E:/Users/BROTHER COMPUTER/Desktop/research/Cleaned_Data/master_clean.csv")
```

```r
before_war <- filter(master_csv, war_ru == 0)
war_r <- filter(master_csv, war_ru == 1)
war_i <- filter(master_csv, war_ii == 1)
before_war <- group_by(before_war, commodity) %>% 
summarise(
    avg_price = mean(Average_price, na.rm = TRUE),
    std = sd(Average_price, na.rm = TRUE),
    min = min(Average_price, na.rm = TRUE),
    max = max(Average_price, na.rm = TRUE)
)
war_r <- group_by(war_r , commodity) %>% 
summarise(
    avg_price = mean(Average_price, na.rm =  TRUE),
    std = sd(Average_price, na.rm =  TRUE),
    min = min(Average_price, na.rm =  TRUE),
    max = max(Average_price, na.rm =  TRUE)
)
war_i <- group_by(war_i , commodity) %>% 
summarise(
    avg_price = mean(Average_price, na.rm =  TRUE),
    std = sd(Average_price, na.rm =  TRUE),
    min = min(Average_price, na.rm =  TRUE),
    max = max(Average_price, na.rm =  TRUE)
)

write.csv(before_war, "E:/Users/BROTHER COMPUTER/Desktop/research/Cleaned_Data/before_war_description.csv")
write.csv(war_r, "E:/Users/BROTHER COMPUTER/Desktop/research/Cleaned_Data/war_r_description.csv")
write.csv(war_i, "E:/Users/BROTHER COMPUTER/Desktop/research/Cleaned_Data/war_i_description.csv")
```
