```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("master_clean.csv")
df['Date']=pd.to_datetime(df['Date'])

plot = sns.relplot(data=df, x="Date", y="monthly_average", col="commodity",
                col_wrap=2, kind="line", height=3, aspect=2,
                facet_kws={"sharey": False})

for ax in plot.axes.flat:
    ax.axvline(pd.Timestamp("2022-02-01"), color="red")      
    ax.axvline(pd.Timestamp("2024-04-01"), color="orange")   

plt.show()
```
